```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate synthetic data (100 samples, 2 features)
np.random.seed(42)
X = 2 * np.random.rand(100, 2)
true_w = np.array([3.5, -1.8])
true_b = 4.2
y = np.dot(X, true_w) + true_b + np.random.randn(100) * 0.2 # with slight noise

# 2. Initialize parameters
m, n = X.shape
w = np.zeros(n)
b = 0.0
learning_rate = 0.1
epochs = 500
cost_history = []

# 3. Gradient Descent Loop
for i in range(epochs):
    # TODO 1: Calculate predictions (y_hat) using np.dot
    y_hat = np.dot(X, w) + b

    # TODO 2: Calculate MSE Cost (1 / 2m) * sum((y_hat - y)^2)
    cost = (1 / (2 * m)) * np.sum((y_hat - y) ** 2)
    cost_history.append(cost)

    # TODO 3: Calculate gradients dw and db
    dw = (1 / m) * np.dot(X.T, (y_hat - y))
    db = (1 / m) * np.sum(y_hat - y)

    # TODO 4: Update weights w and bias b
    w = w - learning_rate * dw
    b = b - learning_rate * db

print(f"Learned Weights: {w.round(2)} | True Weights: {true_w}")
print(f"Learned Bias: {round(b, 2)} | True Bias: {true_b}")

# Plot Cost Convergence
plt.plot(cost_history)
plt.title("Cost (J) vs. Iterations")
plt.xlabel("Iteration")
plt.ylabel("MSE Cost")
plt.show()
```
